```python
import pandas as pd
import plotly.express as px


car_data = pd.read_csv('vehicles_us.csv') 

print(car_data.head(10))
print()
car_data.info()
print()
print(car_data.isna().sum())

```

```python
car_data["odometer"] = car_data["odometer"].fillna("unknown")
car_data["is_4wd"] = car_data["is_4wd"].fillna(0).astype(bool)
car_data["cylinders"] = car_data["cylinders"].fillna(0).astype("int")
car_data["date_posted"] = pd.to_datetime(car_data["date_posted"], format="%Y-%m-%d")
car_data.info()
```

```python
print(car_data.groupby("cylinders")["model"].count())
print()
print(car_data.groupby("is_4wd")["model"].count())
print()
print(car_data.groupby("transmission")["model"].count())
print()
print(car_data.groupby("fuel")["model"].count())
```
